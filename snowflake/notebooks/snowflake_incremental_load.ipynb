```python
from pyspark.sql import functions as F

dbutils.widgets.text("catalog_name", "workspace", "Unity Catalog name")
dbutils.widgets.text("source_file_name", "creditcard.csv", "Source file name")
dbutils.widgets.text("batch_id", "batch_001", "Batch ID")

catalog_name = dbutils.widgets.get("catalog_name").strip()
source_file_name = dbutils.widgets.get("source_file_name").strip()
batch_id = dbutils.widgets.get("batch_id").strip()

silver_table = f"{catalog_name}.silver.transactions_silver_incremental"

silver_batch_df = (
    spark.table(silver_table)
    .filter(
        (F.col("batch_id") == batch_id) &
        (F.col("source_file") == source_file_name)
    )
)

print("Silver table:", silver_table)
print("Batch ID:", batch_id)
print("Source file:", source_file_name)
print("Rows to load:", silver_batch_df.count())

display(silver_batch_df.limit(10))
```

```python
sf_options = {
    "host": "zt63149.ap-south-1.aws.snowflakecomputing.com",
    "sfUser": "AKHIL",
    "sfPassword": dbutils.secrets.get(
        scope="fraud-snowflake",
        key="snowflake-password"
    ),
    "sfDatabase": "FRAUD_DB",
    "sfSchema": "RAW",
    "sfWarehouse": "FRAUD_WH",
    "sfRole": "ACCOUNTADMIN"
}

print("Snowflake connection options are ready.")
```

```python
# Ensure the Snowflake RAW table exists before the duplicate-check query runs.
# limit(0) keeps the Silver schema but sends zero transaction rows.

silver_schema_df = (
    spark.table("workspace.silver.transactions_silver_incremental")
    .limit(0)
)

(
    silver_schema_df.write
    .format("snowflake")
    .options(**sf_options)
    .option("dbtable", "TRANSACTIONS")
    .mode("append")
    .save()
)

print("Snowflake RAW.TRANSACTIONS table is ready.")
```

```python
# Count the selected Silver batch.
silver_batch_count = silver_batch_df.count()

if silver_batch_count == 0:
    raise ValueError(
        f"No Silver rows found for batch_id={batch_id}, "
        f"source_file_name={source_file_name}."
    )

# Make parameter values safe for the Snowflake SQL query.
batch_id_sql = batch_id.replace("'", "''")
source_file_sql = source_file_name.replace("'", "''")

# Check whether this exact batch and source file are already loaded.
existing_batch_df = (
    spark.read
    .format("snowflake")
    .options(**sf_options)
    .option(
        "query",
        f"""
        SELECT COUNT(*) AS EXISTING_ROW_COUNT
        FROM FRAUD_DB.RAW.TRANSACTIONS
        WHERE BATCH_ID = '{batch_id_sql}'
          AND SOURCE_FILE = '{source_file_sql}'
        """
    )
    .load()
)

existing_row_count = int(
    existing_batch_df.first()["EXISTING_ROW_COUNT"]
)

print("Silver batch rows:", silver_batch_count)
print("Existing Snowflake rows for this batch:", existing_row_count)

# Load only a new batch. A repeat run is safely skipped.
if existing_row_count == 0:
    (
        silver_batch_df.write
        .format("snowflake")
        .options(**sf_options)
        .option("dbtable", "TRANSACTIONS")
        .mode("append")
        .save()
    )
    print(f"Loaded new batch: {batch_id}")
else:
    print(f"Skipped: batch {batch_id} is already loaded.")

# Reconcile the final Snowflake count for this batch.
loaded_batch_df = (
    spark.read
    .format("snowflake")
    .options(**sf_options)
    .option(
        "query",
        f"""
        SELECT COUNT(*) AS LOADED_ROW_COUNT
        FROM FRAUD_DB.RAW.TRANSACTIONS
        WHERE BATCH_ID = '{batch_id_sql}'
          AND SOURCE_FILE = '{source_file_sql}'
        """
    )
    .load()
)

loaded_row_count = int(
    loaded_batch_df.first()["LOADED_ROW_COUNT"]
)

if loaded_row_count != silver_batch_count:
    raise ValueError(
        f"Reconciliation failed. Silver={silver_batch_count}, "
        f"Snowflake RAW={loaded_row_count}."
    )

print(
    f"Reconciliation passed. Silver={silver_batch_count}, "
    f"Snowflake RAW={loaded_row_count}."
)
```
